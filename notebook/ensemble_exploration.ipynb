```python
# Import required libraries

import pandas as pd
import numpy as np

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split

from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier

from sklearn.metrics import accuracy_score, f1_score, classification_report
```

```python
# Load the Breast Cancer dataset

data = load_breast_cancer()

# Create DataFrame
X = pd.DataFrame(
    data.data,
    columns=data.feature_names
)

# Target variable
y = data.target

# Check for missing values
print("Missing Values:")
print(X.isnull().sum())

# Split the dataset
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nDataset Loaded Successfully!")

print("Training Samples:", X_train.shape[0])
print("Testing Samples:", X_test.shape[0])
```

```python
# Train the baseline Decision Tree model

decision_tree = DecisionTreeClassifier(random_state=42)

decision_tree.fit(X_train, y_train)

# Make predictions
dt_predictions = decision_tree.predict(X_test)

# Evaluate performance
dt_accuracy = accuracy_score(y_test, dt_predictions)
dt_f1 = f1_score(y_test, dt_predictions)

print("Decision Tree Performance")
print("-------------------------")
print(f"Accuracy : {dt_accuracy:.4f}")
print(f"F1-Score : {dt_f1:.4f}")

print("\nClassification Report:\n")
print(classification_report(y_test, dt_predictions))
```

```python
# Train Random Forest model

random_forest = RandomForestClassifier(random_state=42)
random_forest.fit(X_train, y_train)

rf_predictions = random_forest.predict(X_test)

rf_accuracy = accuracy_score(y_test, rf_predictions)
rf_f1 = f1_score(y_test, rf_predictions)


# Train Gradient Boosting model

gradient_boost = GradientBoostingClassifier(random_state=42)
gradient_boost.fit(X_train, y_train)

gb_predictions = gradient_boost.predict(X_test)

gb_accuracy = accuracy_score(y_test, gb_predictions)
gb_f1 = f1_score(y_test, gb_predictions)


# Compare all models

comparison = pd.DataFrame({
    "Model": [
        "Decision Tree",
        "Random Forest",
        "Gradient Boosting"
    ],
    "Accuracy": [
        dt_accuracy,
        rf_accuracy,
        gb_accuracy
    ],
    "F1-Score": [
        dt_f1,
        rf_f1,
        gb_f1
    ]
})

print(comparison)
```

# Results & Conclusion

## Model Performance Summary

Three machine learning models were trained and evaluated on the Breast Cancer Wisconsin Dataset:

- Decision Tree
- Random Forest
- Gradient Boosting

The performance comparison was based on Accuracy and F1-Score.

### Findings

- The Decision Tree served as the baseline model and achieved good classification performance.
- The Random Forest model improved prediction accuracy by combining multiple decision trees, making it more robust and reducing overfitting.
- The Gradient Boosting model also performed very well by sequentially correcting the errors of previous models.

Among the three models, the ensemble methods (Random Forest and Gradient Boosting) generally produced better performance than the single Decision Tree model.

## Conclusion

This experiment demonstrates that ensemble learning techniques provide more accurate and reliable predictions compared to a single Decision Tree.

Random Forest is effective in reducing overfitting through bagging, while Gradient Boosting improves performance by learning from previous mistakes.

Therefore, ensemble methods are preferred for many real-world classification problems because they provide better generalization and prediction accuracy.
